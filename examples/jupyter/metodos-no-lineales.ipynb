# Ejemplos de métodos numéricos usando chuspanumeric en Jupyter

Notebook de ejemplo ejecutando la librería chuspanumeric mediante un kernel de JavaScript.

## Configuración

Para ejecutar los ejemplos se requiere Jupyter con un kernel JavaScript instalado como IJavascript.

```javascript
import { biseccion, simpson, lagrange } from 'chuspanumeric';
```

## Método de bisección (ecuaciones no lineales)

El método de bisección permite encontrar raíces de funciones dividiendo repetidamente un intervalo.

```javascript
const funcion = (x) => x * x - 4;
const raiz = biseccion(funcion, 0, 5);
console.log(raiz);
```

## Método de Simpson (integración numérica)

El método de Simpson aproxima el valor de una integral definida usando interpolación polinómica.

```javascript
const integral = simpson(x => x*x, 0, 2, 10);
console.log(integral);
```

## Método de interpolación de Lagrange

La interpolación de Lagrange permite construir una función aproximada usando puntos conocidos.

```javascript
const puntosX = [0, 1, 2];
const puntosY = [1, 3, 5];
const valor = lagrange(puntosX, puntosY, 1.5);
console.log(valor);
```
